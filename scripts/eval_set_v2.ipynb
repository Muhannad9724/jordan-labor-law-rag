# Eval set v2 — rebuilt to spec

Replaces the 40-question `eval/eval_draft.jsonl` (v1) with 60 questions built to an explicit brief rather than sampled ad-hoc.

## The brief

| source file | language | count | scope |
|---|---|---|---|
| `scl_articles_ar` | ar | 15 | chapters 4–8 only |
| `scl_articles_en` | en | 15 | chapters 4–8 only |
| `worklaw_ar_articles` | ar | 13 | what actually matters to a worker |
| `worklaw_articles_en` | en | 13 | what actually matters to a worker |
| `worklaw_prof_ar_articles` | ar | 2 | what actually matters to a worker |
| `worklaw_prof_articles_en` | en | 2 | what actually matters to a worker |

**No cross-lingual questions.** Every question is answered from a source document in the same language it was asked in — `question_language == answer_language` for all 60 rows. v1 had 8 cross-lingual pairs; they are gone by design.

**No institutional trivia.** v1 spent questions on things a worker would never ask (governance bodies, ministry structure, inspector powers). v2 excludes them. What survives is money, leave, injury, termination, and pensions.

**Multi-article questions are deliberate.** 8 of the 60 require combining several articles into one answer — most notably "what types of leave is a worker entitled to", which spans annual (61), sick (65), additional (66) and maternity (70). v1 had almost none of these, so nothing in it tested whether the system can *assemble* an answer rather than look one up.

## Register: formal in both languages

Questions are written in **formal register in both Arabic and English** — Modern Standard Arabic, not Jordanian colloquial, and neutral formal English rather than the conversational first-person phrasing v1 used ("I got hurt at work, what does social security cover?").

This is a change of direction from v1, whose Phase 3 notes explicitly chose "plain, how-a-worker-would-ask phrasing". Two reasons for the switch:

1. **Register parity between the language arms.** If the Arabic questions were colloquial and the English conversational-but-standard, any EN-vs-AR performance gap would be confounded by register rather than measuring language. Holding register constant across both arms is what makes the comparison mean something.
2. **Register distance from the corpus.** The source documents are formal legal Arabic and formal legal English. Colloquial queries add a second, uncontrolled variable — lexical distance from the corpus — on top of the retrieval question actually being asked.

The tradeoff is real and worth stating: formal questions are **less representative of how a real user will type into the deployed app**. This eval set measures retrieval quality against the corpus, not robustness to casual phrasing. Colloquial-input robustness is a separate thing that deserves its own test set rather than being tangled into this one.

## Three data problems found while building this

These were discovered by reading the knowledge base directly and are worth recording regardless of the eval set.

**1. The SCL `Chapter` column is off by one at every boundary.** Article 41 is labelled *Chapter Five: Maternity* but its content is work-injury-by-third-party (Chapter Four). Same shift at 47, 58, 70. Real content boundaries: Ch4 `23–41`, Ch5 `42–47`, Ch6 `48–58`, Ch7 `59–70`, Ch8 `71–77`. Chapter selection below is by **content**, not by the label. The overall range (23–77) is unaffected.

**2. Professional Work Law article numbers diverge between EN and AR from Article 6 onward.** EN has 13 articles, AR has 12; AR runs one behind (EN Art 7 "professional workers shall be classified into categories" = AR Art 6). This is the downstream effect of the already-known issue that the EN file is the unamended 2019 original while the AR file is amended-to-2024. Each language's questions are therefore labelled with **that language's** numbering — an important correctness detail that a cross-lingual question would have silently gotten wrong.

**3. `Article #` dtype is inconsistent across the knowledge base** — `str` in both SCL files, `int64` in the Labor Law files. `chunks_articles.parquet` normalises everything to `str`, which is what ground truth is validated against below.

No API calls anywhere in this notebook — ground truth comes from reading the source tables, never from running the retriever. Labelling questions with whatever the production pipeline returns would grade the system against itself.

```python
import json
import re
from pathlib import Path

import pandas as pd

PROJECT = Path(r"c:\Users\Mohannad Alsaleh\Desktop\Data Science\Data Science Projects\1st RAG project")
KB = PROJECT / "0. cleaned_data" / "Knowledgebase"

# source file -> (law code used in chunks_articles.parquet, language)
SOURCES = {
    "scl_articles_ar.xlsx":          ("ssl", "ar"),
    "scl_articles_en.xlsx":          ("ssl", "en"),
    "worklaw_ar_articles.xlsx":      ("labor96", "ar"),
    "worklaw_articles_en.xlsx":      ("labor96", "en"),
    "worklaw_prof_ar_articles.xlsx": ("prof19", "ar"),
    "worklaw_prof_articles_en.xlsx": ("prof19", "en"),
}

kb = {}
for fn, (law, lang) in SOURCES.items():
    df = pd.read_excel(KB / fn)
    df["Article #"] = df["Article #"].astype(str).str.strip()   # dtype is str in SCL, int64 in labor -- normalise
    kb[(law, lang)] = df.set_index("Article #")
    print(f"{fn:32s} -> {law:8s} {lang}  {len(df):4d} articles")
```

## Assemble, auto-fill, and validate

`source_article_excerpt` is pulled from the knowledge base rather than typed by hand, so it cannot drift from the real text. The validation cell is the point of this notebook: it fails loudly if any ground-truth article does not exist in the index the retriever actually searches.

```python
BLOCKS = [("en", SCL_EN), ("ar", SCL_AR), ("en", LABOR_EN), ("ar", LABOR_AR),
          ("en", PROF_EN), ("ar", PROF_AR)]

# Enumerations get their own line. A note listing seven conditions as one semicolon-separated
# paragraph is technically complete but unreviewable -- a human checking it against the law
# cannot see at a glance whether an item is missing. Applied here rather than baked into each
# note by hand so the rule is one place and uniform.
# Every pattern requires the marker to follow sentence-final punctuation, so a newline can
# never be inserted into the middle of a clause.
_BREAKS = [
    r"(?<=[.:؛])\s+(?=\([a-h]\)\s)",          # (a) (b) (c) (d) (h)
    r"(?<=[.:؛])\s+(?=\([أ-ي]\)\s)",  # (أ) (ب) (ج) (د)
    r"(?<=[.:؛])\s+(?=Article\s+\d+\s*(?:--|:))",
    r"(?<=[.:؛])\s+(?=المادة\s+\d+\s*:)",
    r"(?<=[.:؛])\s+(?=\d\.\s)",                # 1. 2. 3.
]

def format_note(note: str) -> str:
    for pat in _BREAKS:
        note = re.sub(pat, "\n", note)
    return note

def excerpt(law, lang, art, n=200):
    """First n characters of the real source article, whitespace-collapsed."""
    row = kb[(law, lang)].loc[art]
    return re.sub(r"\s+", " ", str(row["Content"])).strip()[:n]

rows = []
for lang, block in BLOCKS:
    for qtext, law, arts, cat, diff, note in block:
        rows.append({
            "question_id": f"q{len(rows) + 1:02d}",
            "question_text": qtext,
            "question_language": lang,
            "answer_law": law,
            "answer_article_numbers": arts,
            "answer_language": lang,          # no cross-lingual questions, by design
            "question_category": cat,
            "labeling_note": format_note(note),
            "difficulty": diff,
            "n_articles": len(arts),
            "source_article_excerpt": " || ".join(excerpt(law, lang, a) for a in arts),
        })

ev = pd.DataFrame(rows)
multi_line = (ev.labeling_note.str.count("\n") > 0).sum()
print(f"{len(ev)} questions | {multi_line} notes formatted as multi-line lists")
ev.head(3)[["question_id", "answer_law", "answer_article_numbers", "question_language"]]
```

```python
chunks = pd.read_parquet(PROJECT / "0. cleaned_data" / "chunks_articles.parquet")
index_keys = set(zip(chunks.law, chunks.language, chunks.article_number.astype(str)))

failures = []

# 1. every ground-truth article must exist in the index the retriever searches
for r in ev.itertuples(index=False):
    for a in r.answer_article_numbers:
        if (r.answer_law, r.answer_language, a) not in index_keys:
            failures.append(f"{r.question_id}: ({r.answer_law}, {r.answer_language}, art {a}) NOT in chunk index")

# 2. no cross-lingual questions
for r in ev.itertuples(index=False):
    if r.question_language != r.answer_language:
        failures.append(f"{r.question_id}: cross-lingual ({r.question_language} -> {r.answer_language})")

# 3. SCL questions must sit in chapters 4-8, i.e. articles 23-77 by content
for r in ev.itertuples(index=False):
    if r.answer_law == "ssl":
        for a in r.answer_article_numbers:
            if not (a.isdigit() and 23 <= int(a) <= 77):
                failures.append(f"{r.question_id}: SCL art {a} outside chapters 4-8 (23-77)")

# 4. counts must match the brief exactly
SPEC = {("ssl", "ar"): 15, ("ssl", "en"): 15, ("labor96", "ar"): 13,
        ("labor96", "en"): 13, ("prof19", "ar"): 2, ("prof19", "en"): 2}
actual = ev.groupby(["answer_law", "answer_language"]).size().to_dict()
for key, want in SPEC.items():
    got = actual.get(key, 0)
    if got != want:
        failures.append(f"count {key}: expected {want}, got {got}")

# 5. ids unique, no empty text
if ev.question_id.duplicated().any():
    failures.append("duplicate question_id")
if (ev.question_text.str.strip() == "").any() or (ev.labeling_note.str.strip() == "").any():
    failures.append("empty question_text or labeling_note")

# 6. no vague deferrals in the ground truth.
#    A note that says "as the article specifies" instead of stating the rule is useless as
#    ground truth: a model that gives the real rule cannot be credited for it, and a model
#    that punts the same way scores as correct. 13 of 60 notes failed this on the first pass
#    -- caught in review, not by the earlier checks -- so it is now enforced.
HEDGES = ["the article specifies", "the article lists", "the article sets", "sets out",
          "listed cases", "listed protected", "further cases", "the stated limits",
          "الواردة في المادة", "المفصّلة في", "المفصلة في", "التي تعددها المادة",
          "المحددة في المادة", "وفق القواعد", "الحالات الأخرى",
          "وفق ما تفصّله", "وفق ما تفصله", "كما تفصّله"]
for r in ev.itertuples(index=False):
    hit = [h for h in HEDGES if h in r.labeling_note]
    if hit:
        failures.append(f"{r.question_id}: ground truth defers instead of stating the rule -> {hit}")

print("FAILURES:", len(failures))
for f in failures:
    print(" ", f)
assert not failures, "validation failed -- do not save"
print("\nall checks passed")
```
